## Question 1
Write a function called `proportion_of_education` which returns the proportion of children in the dataset who had a mother with the education levels equal to less than high school (<12), high school (12), more than high school but not a college graduate (>12) and college degree.

*This function should return a dictionary in the form of (use the correct numbers, do not round numbers):* 
```
    {"less than high school":0.2,
    "high school":0.4,
    "more than high school but not college":0.2,
    "college":0.2}
```

### Solution Explanation

The function parses the raw `NISPUF17.csv` dataset and extracts the column `EDUC1`, which denotes maternal education levels.

- `len(df)` is used to calculate the absolute denominator (total children surveyed).
- `df['EDUC1'].value_counts()` aggregates the counts of every unique education level code into a Series mapping out exact frequency.
- Finally, we compose the requested dictionary by returning the frequencies generated by `counts.get()` divided by the total.

````python
def proportion_of_education():
    import pandas as pd
    
    # Load dataset
    df = pd.read_csv('datasets/NISPUF17.csv')
    total = len(df)
    
    # Extract counts of each distinct education group representation (1=Less than HS, 2=HS, 3=Some college, 4=College)
    counts = df['EDUC1'].value_counts()
    
    # Create proportionality dictionary mapping descriptions to ratios
    return {
        "less than high school": counts.get(1, 0) / total,
        "high school": counts.get(2, 0) / total,
        "more than high school but not college": counts.get(3, 0) / total,
        "college": counts.get(4, 0) / total 
    }
````

````python
assert type(proportion_of_education())==type({}), "You must return a dictionary."
assert len(proportion_of_education()) == 4, "You have not returned a dictionary with four items in it."
assert "less than high school" in proportion_of_education().keys(), "You have not returned a dictionary with the correct keys."
assert "high school" in proportion_of_education().keys(), "You have not returned a dictionary with the correct keys."
assert "more than high school but not college" in proportion_of_education().keys(), "You have not returned a dictionary with the correct keys."
assert "college" in proportion_of_education().keys(), "You have not returned a dictionary with the correct keys."
````
