## Question 3
It would be interesting to see if there is any evidence of a link between vaccine effectiveness and sex of the child. Calculate the ratio of the number of children who contracted chickenpox but were vaccinated against it (at least one varicella dose) versus those who were vaccinated but did not contract chicken pox. Return results by sex. 

*This function should return a dictionary in the form of (use the correct numbers):* 
```
    {"male":0.2,
    "female":0.4}
```

Note: To aid in verification, the `chickenpox_by_sex()['female']` value the autograder is looking for starts with the digits `0.0077`.

````python
def chickenpox_by_sex():
    import pandas as pd
    df = pd.read_csv('assets/NISPUF17.csv')
    vax = df[(df['P_NUMVRC'] >= 1) & (df['HAD_CPOX'].isin([1, 2]))]
    male_vax = vax[vax['SEX'] == 1]
    male_cpox = len(male_vax[male_vax['HAD_CPOX'] == 1])
    male_nocpox = len(male_vax[male_vax['HAD_CPOX'] == 2])
    female_vax = vax[vax['SEX'] == 2]
    female_cpox = len(female_vax[female_vax['HAD_CPOX'] == 1])
    female_nocpox = len(female_vax[female_vax['HAD_CPOX'] == 2])
    return {
        "male": male_cpox / male_nocpox if male_nocpox else 0,
        "female": female_cpox / female_nocpox if female_nocpox else 0
    }
    raise NotImplementedError()
````

````python
assert len(chickenpox_by_sex())==2, "Return a dictionary with two items, the first for males and the second for females."
````
