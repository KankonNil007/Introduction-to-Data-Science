## Question 2

Let's explore the relationship between being fed breastmilk as a child and getting a seasonal influenza vaccine from a healthcare provider. Return a tuple of the average number of influenza vaccines for those children we know received breastmilk as a child and those who know did not.

*This function should return a tuple in the form use the correct numbers:*
```
(2.5, 0.1)
```

### Solution Explanation

This solution calculates conditional means based on two specific criteria (Breastmilk history vs No Breastmilk history).

- `dropna(subset=['P_NUMFLU'])` drops any individual rows that do not have recorded influenza vaccine datums to ensure accurate average calculations.
- `['CBF_01'] == 1` queries rows identifying children who **did** receive breastmilk. We then grab the `P_NUMFLU` column from that subset and aggregate its mean.
- `['CBF_01'] == 2` runs the opposite subset calculation for those who **did not**.
- Finally, we return the two means bundled in a tuple.

````python
def average_influenza_doses():
    import pandas as pd
    df = pd.read_csv('datasets/NISPUF17.csv')
    
    # Remove null vaccines from calculation to avoid biasing average
    valid_flu = df.dropna(subset=['P_NUMFLU'])
    
    # Calculate mean of influenza vaccines for 'yes' breastfed answers (1)
    yes = valid_flu[valid_flu['CBF_01'] == 1]['P_NUMFLU'].mean()
    
    # Calculate mean of influenza vaccines for 'no' breastfed answers (2)
    no = valid_flu[valid_flu['CBF_01'] == 2]['P_NUMFLU'].mean()
    
    return (yes, no)
````

````python
assert len(average_influenza_doses())==2, "Return two values in a tuple, the first for yes and the second for no."
````
