# Pandas DataFrame Data Structure

A **DataFrame** is the core of Pandas. It is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). You can think of it like a spreadsheet or a SQL table.

This notebook covers:
- Creating DataFrames from Series and Dictionaries
- Querying rows and columns using `.loc`
- Transposing DataFrames
- Manipulating columns (dropping and copying)

```python
# Import the Pandas library
import pandas as pd
```

## Creating a DataFrame from Multiple Series

You can build a DataFrame by passing a list of `pd.Series` items. Each Series will represent a row. The index of the DataFrame must be provided separately to label these rows.

```python
# Define three separate Series representing three students' records
record1 = pd.Series({'Name': 'Alice',
                        'Class': 'Physics',
                        'Score': 85})
record2 = pd.Series({'Name': 'Jack',
                        'Class': 'Chemistry',
                        'Score': 82})
record3 = pd.Series({'Name': 'Helen',
                        'Class': 'Biology',
                        'Score': 90})
```

```python
# Create the DataFrame by passing the Series in a list.
# Set custom index names matching the school each student belongs to.
df = pd.DataFrame([record1, record2, record3],
                  index=['school1', 'school2', 'school3'])
df.head()
```

## Creating a DataFrame from a List of Dictionaries

Instead of creating Series objects first, you can directly pass a list of Python dictionaries to initiate the DataFrame.

```python
# Define a list where each element is a dictionary representing a row
students = [{'Name': 'Alice',
              'Class': 'Physics',
              'Score': 85},
            {'Name': 'Jack',
             'Class': 'Chemistry',
             'Score': 82},
            {'Name': 'Helen',
             'Class': 'Biology',
             'Score': 90}]
```

```python
# Create DataFrame. Note that indices do not have to be unique!
df = pd.DataFrame(students, index=['school1', 'school2', 'school1'])

df.head()
```

## Accessing Rows in a DataFrame

To access a row, use `.loc[]` with the index name. 
- If the index is **unique**, Pandas returns a `Series`.
- If the index is **duplicated**, Pandas returns a subset `DataFrame` containing all matching rows.

```python
# Accessing a row with a unique index returns a Series
print(df.loc["school2"])

print(type(df.loc["school2"]))
```

```python
# Accessing a row with a duplicated index returns a DataFrame
print(df.loc["school1"])

print(type(df.loc["school1"]))
```

```python
# You can also query a specific row and a specific column simultaneously using `.loc[row, column]`
df.loc["school1", "Name"]
```

## Transposing and Column Access

You can transpose a DataFrame (swap rows and columns) using the `.T` attribute.

```python
# Transpose the DataFrame
df.T
```

```python
# After transposing, what was originally a column ('Name') is now an index.
df.T.loc["Name"]
```

## Extracting Columns

The standard and most efficient way to access a column in Pandas is using the bracket operator `[]` directly on the DataFrame.

```python
# Direct bracket accessor gets the entire 'Name' column as a Series
df["Name"]
```

```python
# Chaining operations: first query rows labeled 'school1', then get the 'Name' column from the subset
df.loc["school1"]["Name"]
```

```python
# As seen before, verifying the object types returned
print(type(df.loc["school1"]["Name"]))
print(type(df.loc["school1"]))
```

```python
# .loc[rows, columns] allows fetching all rows for multiple specific columns safely
df.loc[:, ["Name", "Score"]]
```

## Dropping Data

You can use `.drop()` to remove rows or columns. By default, Pandas operates on an immutable baseline—meaning it doesn't change the original DataFrame unless you explicitly use `inplace=True` or reassign the variable.

```python
# Drop the row labeled 'school1'. 
# This returns a NEW DataFrame without modifying `df`.
df.drop("school1")
```

```python
# Printing the original DataFrame proves that 'school1' is still there.
df
```

```python
# Make a separate copy of the DataFrame
cpydf = df.copy()

# `inplace=True` forces the change to persist in `cpydf`.
# `axis=1` tells Pandas to drop a COLUMN instead of a row.
cpydf.drop("Name", inplace=True, axis=1)
cpydf
```

```python
# Alternatively, the `del` keyword can be used to permanently remove a column in-place instantly.
del cpydf["Class"]
cpydf
```

## Adding New Columns

Adding columns is as simple as defining a new key in a dictionary, using bracket notation.

```python
# Add a new column 'ClassRanking' and set all its values to None to be filled later
df['ClassRanking'] = None
df
```
