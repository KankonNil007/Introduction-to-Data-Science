```python
import pandas as pd
```

```python
record1 = pd.Series({'Name': 'Alice',
                        'Class': 'Physics',
                        'Score': 85})
record2 = pd.Series({'Name': 'Jack',
                        'Class': 'Chemistry',
                        'Score': 82})
record3 = pd.Series({'Name': 'Helen',
                        'Class': 'Biology',
                        'Score': 90})
```

```python
df = pd.DataFrame([record1, record2, record3],
                  index=['school1', 'school2', 'school3'])
df.head()
```

```python
students = [{'Name': 'Alice',
              'Class': 'Physics',
              'Score': 85},
            {'Name': 'Jack',
             'Class': 'Chemistry',
             'Score': 82},
            {'Name': 'Helen',
             'Class': 'Biology',
             'Score': 90}]
```

```python
df = pd.DataFrame(students, index=['school1', 'school2', 'school1'])

df.head()
```

```python
print(df.loc["school2"]) # Aaccessing Rows

print(type(df.loc["school2"]))
```
