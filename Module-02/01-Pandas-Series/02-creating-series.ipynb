# Advanced Ways to Create Pandas Series

This notebook explores various ways to instantiate a `pandas.Series`. You will learn how to create a Series using dictionaries, lists of tuples, and how to define custom indices.

```python
# Import the pandas library for data manipulation
import pandas as pd
```

## Creating a Series from a Dictionary

When a Pandas Series is created from a dictionary, the dictionary's keys seamlessly become the index of the Series, and the dictionary's values become the data points.

```python
# A dictionary mapping student names to the subjects they study
subjects = {
    "Alice": "Physics",
    "John": "Chemistry",
    "Molly": "Mathematics"
}

# Create Series from the dictionary
ds = pd.Series(subjects)

# Display the result; notice the keys are now the index
print(ds)
```

## Accessing the Index of a Series

You can isolate and inspect the index of any Series using the `.index` attribute. This is useful when you need to iterate over indices or check for their presence.

```python
# Print the index labels of the Series
print(ds.index)
```

## Creating a Series from a List of Tuples

You can also pass a list of tuples to a Series. The entire tuple elements are stored sequentially as data, and index falls back to the default `0, 1, 2...` format.

```python
# Create a list of tuples containing a name and a secondary attribute
students = [("Alice","Brown"), ("Jack", "White"), ("Molly", "Green")]

# Convert list of tuples into a Series
ds = pd.Series(students)

print(ds)
```

## Explicitly Defining the Index

Instead of letting Pandas auto-generate integers or dictate indices from dict keys, you can manually dictate the index via the `index=` parameter during instantiation.

```python
# Provide custom index labels while defining the data list
ds = pd.Series(["Alice", "Molly", "Jack"], index=["Physics", "Chemistry", "Mathematics"])

print(ds)
```

## Supplying an Incomplete Index to Dict-Based Series

If you pass a custom list of indices when generating a Series from a dictionary, Pandas filters the dictionary based on your provided indices.
- Matches fetch their dictionary values.
- Missing combinations are assigned `NaN`.

```python
subjects = {
    "Alice": "Physics",
    "John": "Chemistry",
    "Molly": "Mathematics"
}

# Supplying specific keys via index parameter.
# 'Sam' doesn't exist in dict so it evaluates to NaN, and 'John' is dropped.
ds = pd.Series(subjects, index=["Alice", "Molly", "Sam"])

print(ds)
```
