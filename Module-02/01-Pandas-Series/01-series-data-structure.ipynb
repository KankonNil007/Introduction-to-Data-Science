# Introduction to Pandas Series

This notebook provides an introduction to the **Pandas Series** data structure. A Series is a one-dimensional array-like object containing an array of data and an associated array of data labels, called its index. 

**Purpose**: To demonstrate how to initialize Series from Python lists containing strings, integers, and missing values, and how Pandas infers data types under the hood.

```python
# Import the pandas library. It is widely used for data manipulation and analysis.
import pandas as pd
```

## Creating a Series of Strings

When creating a Pandas Series from a list of strings, Pandas infers the type of the elements and stores them as `object`.

```python
# Define a list of strings
students = ["Kankon", "David", "Jack"]

# Create the Pandas Series from the students list
studSeries = pd.Series(students)

# Display the Series
print(studSeries)
```

## Creating a Series of Integers

Unlike strings which are treated as `object`, when you create a Series containing only whole numbers, Pandas optimizes efficiency by storing them as `int64`.

```python
# Define a list of integers
numbers = [56, 78, 98]

# Create the Series; Pandas automatically detects the integers and sets dtype to int64
numSeries = pd.Series(numbers)

print(numSeries)
```

## Handling Missing Values in a Series of Strings

If you pass missing data like `None` inside a list of strings, Pandas simply represents it as a `None` type object, preserving the `object` data type.

```python
# Create a list of strings but explicitly include a None value
students = ["Kankon", "David", None]

# The resulting Series still has dtype: object
studSeries = pd.Series(students)

print(studSeries)
```

## Handling Missing Values in a Series of Numbers

Unlike objects, Pandas requires a special floating-point representation `NaN` (Not a Number) to represent missing numeric data efficiently. When it encounters `None` in a list of integers, it automatically casts the entire Series to `float64` to accommodate the `NaN`.

```python
# Create a list of numbers with a None value
numbers = [56, 78, None]

# Notice that the integers have been converted to floats, and None was converted to NaN
numSeries = pd.Series(numbers)

print(numSeries)
```

## Exploring `NaN` and `None` logic

`NaN` (Not a Number) behaves differently than standard Python `None` values. For instance, `NaN` is not equal to `None`, and effectively, `NaN` is also not considered mathematically equal to itself.

```python
import numpy as np

# NaN is not equal to None
print(np.nan == None)

# NaN is essentially undefined data, so it is not even equal to itself
print(np.nan == np.nan)
```

### How to check for missing numbers properly
Since standard equality checks (`==`) fail on `NaN`, you must use a specialized function, like `np.isnan()`, to identify missing variables effectively.

```python
# Using np.isnan() is the correct way to test for a NaN value
print(np.isnan(np.nan))
```
