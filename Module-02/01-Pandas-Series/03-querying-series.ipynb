# Querying a Pandas Series

Once you have a Pandas Series, querying allows you to easily fetch individual or multiple data points. This notebook illustrates positional indexing (`.iloc`), label-based indexing (`.loc`), and nuances like integer indices.

```python
# Import pandas library
import pandas as pd
```

## Setting up the Series

First, we will create a Series from a dictionary to demonstrate querying based on textual custom index labels.

```python
# Dictionary matching student to subject
students_classes = {'Alice': 'Physics',
                   'Jack': 'Chemistry',
                   'Molly': 'English',
                   'Sam': 'History'}

ds = pd.Series(students_classes)
print(ds)
```

## Best Practices for Querying: `iloc` and `loc`

To retrieve values from a Series safely, you should use explicit attributes:
- **iloc**: Looks up values by their numerical positional order (0, 1, 2...).
- **loc**: Looks up values using the explicit index labels (like "Molly").

```python
# Indexing by numerical position (iloc)
# Note: 3 corresponds to the 4th element (Sam -> History)
print(ds.iloc[3])

# Indexing directly by index names/labels (loc)
print(ds.loc["Molly"])
```

## Implicit Indexing Constraints (`[]` brackets) 

While direct bracket lookup like `ds["Molly"]` works, using positional indices directly like `ds[3]` will issue a `FutureWarning` in newer Pandas versions, as Pandas is shifting away from overlapping integer keys to prevent confusion between labels and positions.

```python
# Indexing Shortcuts - (recommended not to use to avoid warnings/unpredictability)
print(ds[3])
print(ds["Molly"])
```

## Indexing with Integer Labels

If explicit labels assigned to the Series are actually integers (e.g. `99, 100`), direct bracket indexing behaves differently and assumes label-lookup, solidifying the need to prefer `.loc` and `.iloc`.

```python
# Creating a new Series where dictionary keys (thus labels) are integers
class_code = {99: 'Physics',
              100: 'Chemistry',
              101: 'English',
              102: 'History'}
ds = pd.Series(class_code)

print(ds)
```

```python
# Use iloc for purely positional access despite labels being sequential integers
print(ds.iloc[2]) # Grabs string at index 2 (101: English)

# Use loc to specify the literal integer label
print(ds.loc[99]) # Grabs string corresponding to label 99 (Physics)
```
