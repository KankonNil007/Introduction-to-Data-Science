# Vectorized Operations in Pandas

This notebook demonstrates why Pandas (and its underlying NumPy architecture) is highly efficient for data manipulation. We will compare slow, traditional Python loops against fast, optimized **vectorized operations** using built-in functions and operators.

```python
import pandas as pd
```

## Calculating Averages: The General Python Method

Traditionally, to find an average, you would iterate through every item, sum them up, and divide by the length of the list. Here, we calculate the average grade of four individuals.

```python
# Using a standard for-loop to sum values
grades = pd.Series([90, 80, 50, 60])

total = 0
for i in grades:
    total += i

average = (total / len(grades))
print(average)
```

## Calculating Averages: The Vectorized Method

With NumPy and Pandas, operations can be **vectorized**. A vectorized function like `np.sum()` computes the summation at the C-level much faster than a Python loop.

```python
import numpy as np

# Using NumPy's highly optimized sum function instead of a loop
total = np.sum(grades)

average = (total / len(grades))
print(average)
```

## Benchmarking Speed: Loop vs Vectorization

To truly see the difference, let's create a large Series of 10,000 random integers and use Jupyter's `%%timeit` magic command to see how long each method takes.

```python
# Create 10,000 random integers between 0 and 1000
numbers = pd.Series(np.random.randint(0,1000,10000))

# Display the last 5 numbers
numbers.tail()
```

### Speed Test 1: For-loop

```python
%%timeit -n 100
total = 0
for number in numbers:
    total+=number

total/len(numbers)
```

### Speed Test 2: NumPy Vectorization

Notice how `np.sum()` is significantly faster (measured in microseconds instead of milliseconds).

```python
%%timeit -n 100
total = np.sum(numbers)
total/len(numbers)
```

## Broadcasting

Another feature of vectorization is **broadcasting**. If we want to add `2` to every number, we don't need a loop, we can just add `2` directly to the Series.

```python
# The original first 5 numbers
numbers.head()
```

```python
# The scalar '2' is broadcasted across all 10,000 elements simultaneously
numbers += 2
numbers.head()
```

### Broadcasting vs `.at` iteration

This is vastly superior to iterating through the `.items()` and manually mutating items using the `.at[]` accessor.

```python
# Modifying the dataset through manual iteration (Slow)
for label, value in numbers.items():
    numbers.at[label] = value+2

numbers.head()
```

### Final Benchmark: Mutating Items


```python
%%timeit -n 10
# Slower manual update using a loop for 1,000 elements
s = pd.Series(np.random.randint(0,1000,1000))
for label, value in s.items():
    s.at[label]= value+2
```

```python
%%timeit -n 10
# Faster update using broadcasting
s = pd.Series(np.random.randint(0,1000,1000))
s+=2
```
