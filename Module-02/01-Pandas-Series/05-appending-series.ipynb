# Appending Data to a Pandas Series

This notebook covers how to expand an existing `pandas.Series`. It demonstrates appending individual items using `.loc` and combining multiple Series objects using the `pd.concat` function, emphasizing handling duplicate index labels.

```python
# Import pandas for data manipulation
import pandas as pd
```

## Appending a Single Element

You can add a new item to a Series in-place by assigning a value to a new index label using `.loc[]`. Since Pandas Series can mix types for index labels, you can append a string-labeled item to an integer-indexed Series.

```python
# Create a standard integer-indexed Series
ds = pd.Series([1, 2, 3])

# Append a new entry with the label 'History' and value 104
ds.loc["History"] = 104

print(ds)
```

## Appending Multiple Elements using `concat`

If you want to append an entire Series to another, using `pd.concat()` is the preferred method rather than adding items one by one. This returns a **new** Series object.

```python
# Base Series representing student subject assignments
students_classes = pd.Series({'Alice': 'Physics',
                   'Jack': 'Chemistry',
                   'Molly': 'English',
                   'Sam': 'History'})
print(students_classes)
```

```python
# Second Series representing multiple classes for one student.
# Note: In Pandas, index labels do not have to be unique!
kelly_classes = pd.Series(['Philosophy', 'Arts', 'Math'], index=['Kelly', 'Kelly', 'Kelly'])
print(kelly_classes)
```

### Executing the Concatenation

We pass a list of the Series objects to `pd.concat()`. This operation combines them vertically.

```python
# Concatenate the two Series. This creates a highly optimized new object.
allStudClasses = pd.concat([students_classes, kelly_classes])

print(allStudClasses)
```

### Original Data is Preserved

Unlike the `.loc` assignment which modifies the Series in-place, `pd.concat()` returns a new object, leaving the originals unmodified.

```python
# Prove that the original base Series is completely unchanged
print(students_classes)
```

### Querying Duplicate Indices

When querying an index that appears multiple times in the Series, Pandas will return a Series combining all matching values, rather than just returning a single string/scalar item.

```python
# Querying 'Kelly' returns all 3 elements associated with the 'Kelly' index label
print(allStudClasses.loc["Kelly"])
```
