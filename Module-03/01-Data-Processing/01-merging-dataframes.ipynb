# Merging DataFrames

This notebook demonstrates how to combine different DataFrames. In pandas, `pd.merge()` is highly versatile and allows for various types of SQL-like joins such as outer (union), inner (intersection), left, and right joins.

```python
# Import pandas for data manipulation
import pandas as pd
```

### Creating Sample DataFrames

We start by initializing two DataFrames: one for university staff and one for students. We explicitly set their indices to the 'Name' column so that `pd.merge` can use the index as the primary key for joining.

```python
# Create staff DataFrame with roles
staff_df = pd.DataFrame([{'Name': 'Kelly', 'Role': 'Director of HR'},
                         {'Name': 'Sally', 'Role': 'Course liasion'},
                         {'Name': 'James', 'Role': 'Grader'}])

# Set 'Name' as the index
staff_df = staff_df.set_index('Name')

# Create student DataFrame with schools
student_df = pd.DataFrame([{'Name': 'James', 'School': 'Business'},
                           {'Name': 'Mike', 'School': 'Law'},
                           {'Name': 'Sally', 'School': 'Engineering'}])

# Set 'Name' as the index here as well
student_df = student_df.set_index('Name')
```

Let's view the `student_df` to see its structure:

```python
# Display student DataFrame
student_df
```

Similarly, let's view the `staff_df`:

```python
# Display staff DataFrame
staff_df
```

### Outer Join (Set Union)

An outer join brings together all rows from both DataFrames. If a key is missing in one DataFrame, the missing values will be filled with `NaN` (Not a Number).

```python
# Perform an outer join to get all records from both DataFrames
# left_index=True and right_index=True indicate we want to use the indices as join keys
pd.merge(staff_df, student_df, how='outer', left_index=True, right_index=True)
```

### Inner Join (Set Intersection)

An inner join only keeps the rows whose index is present in *both* DataFrames.

```python
# Perform an inner join to get only records that exist in BOTH DataFrames
pd.merge(staff_df, student_df, how='inner', left_index=True, right_index=True)
```

### Left Join

A left join returns all rows from the Left DataFrame (`staff_df`), and attaches data from the Right DataFrame where the keys match. Unmatched keys from the right are dropped.

```python
# Left join: get all records of staff, and any extra information from students if they are also students
pd.merge(staff_df, student_df, how='left', left_index=True, right_index=True)
```

### Right Join

A right join is the opposite of a left join. It returns all rows from the Right DataFrame (`student_df`), attaching data from the Left DataFrame where present.

```python
# Right join: get all records of students, and append staff roles where applicable
pd.merge(staff_df, student_df, how='right', left_index=True, right_index=True)
```

### Merging on Columns (Instead of Indices)

If we do not want to use the index, we can perform a merge on specific columns. First, we use `reset_index()` to turn the indices back into regular columns, and then specify `on='Name'`.

```python
# Reset indices so 'Name' goes back to being a normal column
staff_df = staff_df.reset_index()
student_df = student_df.reset_index()

# Perform a right join on the 'Name' column specifically
pd.merge(staff_df, student_df, how='right', on='Name')
```
