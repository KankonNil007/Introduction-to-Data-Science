# Concatenating DataFrames

Sometimes, instead of merging DataFrames horizontally (joining columns), we want to stack them vertically (adding rows). This is very common when dealing with time-series data or logs that are split across multiple files, where each file represents a different time period but shares the exact same columns. This is called concatenation.

```python
# Import pandas for data processing
import pandas as pd
```

### Loading Data

Let's load the College Scorecard data from different years. Notice that since the files are large and have many columns, we pass `low_memory=False` to pandas to ensure it reads the column types correctly without running out of memory while guessing.

```python
# Read College Scorecard data for academic years starting in 2009, 2010, and 2012
# Setting low_memory=False ensures that pandas reads all rows to infer data types correctly
df_2009 = pd.read_csv("datasets/college-scorecard/MERGED2009_10_PP.csv", low_memory=False)
df_2010 = pd.read_csv("datasets/college-scorecard/MERGED2010_11_PP.csv", low_memory=False)
df_2012 = pd.read_csv("datasets/college-scorecard/MERGED2012_13_PP.csv", low_memory=False)
```

Preview the first 3 rows of the 2010 dataset. Notice the huge number of columns (1977!):

```python
# Display the top 3 rows
df_2010.head(3)
```

Let's check the number of rows (entries) in each dataset before combining them.

```python
# Check the length (number of records) of each DataFrame
print(len(df_2010))
print(len(df_2009))
print(len(df_2012))
```

### Basic Concatenation

We pass a list of the DataFrames to `pd.concat()`. This sticks them together row by row. If you sum the numbers from above (7414 + 7149 + 7793), you get 22356, which matches the total row count of the concatenated DataFrame!

```python
# Put the dataframes in a list so we can pass them to the concat function
frames = [df_2009, df_2010, df_2012]

# Concatenate them vertically. Pandas automatically aligns columns with matching names.
pd.concat(frames)
```

### Concatenation with Keys

A problem with simple concatenation is that we lose the origin of the data. To fix this, `pd.concat` provides a `keys` parameter which creates a hierarchical index (MultiIndex) that labels the origin of each row.

```python
# Concatenate the DataFrames but pass a 'keys' list to label which year each chunk belongs to.
# Note: the keys below label the chunks as 2011, 2012, and 2013 respectively.
pd.concat(frames, keys=['2011','2012','2013'])
```
