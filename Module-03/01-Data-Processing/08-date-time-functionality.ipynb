# Date and Time Functionality

Pandas provides powerful tools for working with dates and times. This notebook covers Timestamps, Periods, DatetimeIndex, and more.

```python
import pandas as pd
import numpy as np
```

### Timestamps

`pd.Timestamp` represents a single point in time, essentially pandas' equivalent of Python's `datetime`.

```python
pd.Timestamp('9/1/2019 10:05AM')
```

```python
pd.Timestamp(2019, 12, 20, 0, 0)
```

You can access various attributes like year, month, second, or even `isoweekday()`.

```python
pd.Timestamp(2019, 12, 20, 0, 0).isoweekday()
```

```python
pd.Timestamp(2019, 12, 20, 5, 2,23).second
```

### Periods

Unlike a Timestamp, a `pd.Period` represents a specific span of time (e.g., a specific month or day).

```python
pd.Period('1/2016')
```

```python
pd.Period('3/5/2016')
```

You can perform arithmetic on Periods. For example, adding 5 to a monthly period advances it by 5 months.

```python
pd.Period('1/2016') + 5
```

### DatetimeIndex and PeriodIndex

When we use Timestamps as the index of a Series or DataFrame, pandas automatically creates a `DatetimeIndex`.

```python
t1 = pd.Series(list('abc'), [pd.Timestamp('2016-09-01'), pd.Timestamp('2016-09-02'), 
                             pd.Timestamp('2016-09-03')])
t1
```

```python
type(t1.index)
```

Similarly, when we use Periods as the index, it becomes a `PeriodIndex`.

```python
t2 = pd.Series(list('def'), [pd.Period('2016-09'), pd.Period('2016-10'), 
                             pd.Period('2016-11')])
t2
```

```python
type(t2.index)
```

### Converting to Datetime

Often, dates are imported as strings. We can use `pd.to_datetime()` to intelligently convert them into standard Timestamps.

```python
d1 = ['2 June 2013', 'Aug 29, 2014', '2015-06-26', '7/12/16']

ts3 = pd.DataFrame(np.random.randint(10, 100, (4,2)), index=d1, 
                   columns=list('ab'))
ts3
```

```python
ts3.index = pd.to_datetime(ts3.index, format="mixed")
ts3
```

If your date strings follow the European format (Day first), you can pass `dayfirst=True`.

```python
pd.to_datetime('4.7.12', dayfirst=True)
```

### Timedeltas

`pd.Timedelta` represents a duration or difference in times. Subtracting two Timestamps yields a Timedelta.

```python
pd.Timestamp('9/3/2016')-pd.Timestamp('9/1/2016')
```

```python
pd.Timestamp('9/2/2016 8:10AM') + pd.Timedelta('12D 3H')
```

### Date Offsets

Pandas provides date offsets for working with business calendars or specific cadences (e.g., end of the month, week).

```python
pd.Timestamp('9/4/2016').weekday()
```

```python
pd.Timestamp('9/4/2016') + pd.offsets.Week()
```

```python
pd.Timestamp('9/4/2016') + pd.offsets.MonthEnd()
```
