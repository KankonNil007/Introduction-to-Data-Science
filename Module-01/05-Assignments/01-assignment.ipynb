Task: Use regular expressions to extract specific string components based on patterns. E.g., capitalizing conventions to extract proper names.


Find a list of all of all of the names in the following string using regex.

```python
# Goal: Identify every proper capitalized word using a combination of [A-Z] followed by lowercase alphabets
import re
def names():
    simple_string = """Amy is 5 years old, and her sister Mary is 2 years old. 
    Ruth and Peter, their parents, have 3 kids."""

    # YOUR CODE HERE
    return re.findall("[A-Z][a-z]+", simple_string)
    raise NotImplementedError()
```

```python
# Assert basic length testing against final arrays extracted
assert len(names()) == 4, "There are four names in the simple_string"
```
