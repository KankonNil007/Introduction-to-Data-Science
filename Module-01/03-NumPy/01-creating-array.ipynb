# Introduction to NumPy Array Creation
NumPy provides powerful tools for numerical computations. Here we learn how to create multi-dimensional arrays, check their properties, and generate arrays filled with zeros, ones, or random numbers.


```python
# Import the numpy library (conventionally as 'np') and the math module
import numpy as np
import math
```

```python
# Create a 1D NumPy array and check its number of dimensions (ndim)
a = np.array([1, 2, 3])
print(a)
print(a.ndim)
```

```python
# Create a 2D array (matrix) and inspect its shape and data type
b = np.array([[1, 3, 3], [4, 5, 6]])
print(b)
print(b.shape)
print(b.dtype)
```

```python
# Create an array of floats and observe numpy's automatic type casting to float64
c = np.array([2, 4, 3.3, 5.6])
print(c)
print(c.dtype)
```

```python
# Initialize a 2x3 matrix of zeros and a 3x3 matrix of ones
d = np.zeros((2, 3))
print(d)
e = np.ones((3, 3))
print(e)
```

```python
# Generate a 2x3 matrix with random values between 0 and 1
np.random.rand(2,3) # 0 to 1 (random)
```

```python
# Create an array using arange: from 10 to 29 (exclusive) with a step of 2
f = np.arange(10, 30, 2)
print(f)
```

```python
# Create an array of 5 linearly spaced values between 2 and 15
np.linspace(2, 15, 5)
```
