# Array Manipulation and Math
This notebook demonstrates basic operations such as element-wise arithmetic, boolean tracking/masking, dot products, reshaping, and aggregate functions (sum, min, max, mean).


```python
# Import the numpy library and the math module
import numpy as np
import math

```

```python
# Element-wise arithmetic operations on two arrays of the same shape
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

c = a + b
print(c)

c = a - b
print(c)

c = a * b
print(c)

c = a / b
print(c)
```

```python
# Real-world example: convert array of temperatures from Farenheit to Celcius
farenheit = np.array([5, 0, -4, 23, -10])

celcius = (farenheit - 32) * 5/9

print(celcius)
```

```python
# Boolean array operations: apply conditional statements to every element
print(celcius > -20)
print(celcius%2 == 0)
```

```python
# Matrix operations: compare element-wise multiplication vs the matrix dot product (@)
A = np.array([[1, 2], [3, 4]])
B = np.array([[3, 2], [0, 4]])

print(A*B) # Element-wise Multiplication
print(A@B) # Dot Product
```

```python
# Print the shape of matrix A
print(A.shape)
```

```python
# Check data types of integer vs float arrays
arr1 = np.array([1, 2, 3, 4])
print(arr1.dtype)

arr2 = np.array([10.4, 2.7, 9.3, 6.4])
print(arr2.dtype)
```

```python
# Demonstrate type coercion or upcasting: adding integer array to float array yields a float array
arr3 = arr1 + arr2
print(arr3)
print(arr3.dtype)
```

```python
# Utilize aggregate functions to find sum, max, min, and mean
print(arr3.sum())
print(arr3.max())
print(arr3.min())
print(arr3.mean())
```
