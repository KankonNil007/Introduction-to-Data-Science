# Regular Expressions and Anchors
This notebook introduces basic Regex functionality in Python such as `search`, `split`, and `findall` using strings and anchors (like `^`).


```python
# Import the regular expression module
import re
```

```python
# Simple keyword matching using re.search() returning a Match object if successful
text = "This is a good day"

if re.search("good", text):
    print("Wonderful")
else:
    print("Alas!!")
```

```python
# Demonstrate splitting a string by a pattern and extracting all occurrences of a pattern
text = "Amy works diligently. Amy gets good grades. Our student Amy is successful."

print(re.split("Amy", text))

print(re.findall("Amy", text))
```

```python
# Use the caret anchor '^' to check if a specific pattern occurs exactly at the beginning of the string
text = "Amy works diligently. Amy gets good grades. Our student Amy is succesful."

# Lets see if this begins with Amy
re.search("^Amy",text)
```
