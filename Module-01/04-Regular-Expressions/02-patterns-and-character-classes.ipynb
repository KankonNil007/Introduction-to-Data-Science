# Pattern Matching and Character Classes
Learn how to construct regex rules for character classes (`[A-Z]`), exclusion patterns (`[^A]`), and explicitly defined sets and union operations.


```python
# Import the regular expression module
import re
```

```python
# Match purely 'B' or characters in subset '[AB]'
grades = "ABAAAABBCBABACBA"

print(re.findall("B", grades))
print(re.findall("[AB]", grades))
```

```python
# Use ranges like A to C, or specify sequence strings consecutively (like an A immediately followed by B or C)
print(re.findall("[A-C]", grades))
print(re.findall("[A][B-C]", grades))
```

```python
# Provide logical OR union operator '|' to match explicit separate string scenarios
print(re.findall("AB|AC", grades))
```

```python
# Use the caret '^' inside the character class to negate/exclude the characters
print(re.findall("[^A]", grades)) # skips all As
```

```python
# Combined anchor + exclusion: Match only characters at the very beginning that are NOT 'A'
re.findall("^[^A]",grades)
```
