# Basic Types and Sequences
This notebook explores Python's built-in types (strings, numeric, None) and sequences (tuples, lists, dictionaries), along with common operations like slicing and string formatting.


```python
# Check the type of a string
type("Welcome to Kankon's World")
```

```python
# Define a basic addition function
def add_numbers(a, b):
    return a + b

print(add_numbers(1, 2))
```

```python
# Check the type of None
type(None)
```

```python
# Check the type of an integer
type(1)
```

```python
# Check the type of a float
type(1.0)
```

```python
# Check the type of a function
type(add_numbers)
```

```python
# Create a tuple (immutable sequence) and check its type
tuples = (1, 'a', 2, 'b')
type(tuples)
```

```python
# Create a list (mutable sequence) and check its type
lists = [1, 2, 'a', 'b']
type(lists)
```

```python
# Append a new element to the list
lists.append(3.3)
print(lists)
```

```python
# Iterate through elements of the list using a for loop
for item in lists:
    print(item)
```

```python
# Iterate through elements of the list using a while loop and an index
i = 0
while (i != len(lists)):
    print(lists[i])
    i = i + 1
```

```python
# Concatenate two lists
[1, 2] + [3, 4]
```

```python
# Repeat a list's elements 4 times
[1] * 4
```

```python
# Check if an element exists in a list
1 in [1, 2, 3]
```

```python
# Demonstrate string slicing: accessing characters by index ranges
str1 = "This is a string"
print(str1[0])
print(str1[1:4])
print(str1[:4])
print(str1[4:])
print(str1[-1])
print(str1[-4:-2])
```

```python
# Concatenate strings using the '+' operator
fname = "Kankon"
lname = "Mondal"

print(fname + ' ' + lname)
```

```python
# Split a string by space and extract the first and last names
fname = "Robert Chiris Rowan Simpton".split(" ")[0]
lname = "Robert Chiris Rowan Simpton".split(" ")[-1]
print(fname)
print(lname)
```

```python
# Create a dictionary (key-value pairs) and check its type
dicts = {
    "University": "Harvard",
    "College": "Imperial"
}
type(dicts)
```

```python
# Iterate over the keys of the dictionary
for keys in dicts:
    print(keys)
```

```python
# Iterate over the values of the dictionary
for values in dicts.values():
    print(values)
```

```python
# Iterate over the key-value pairs (items) of the dictionary
for keys, values in dicts.items():
    print(keys)
    print(values)
```

```python
# Unpack a list into multiple variables
x = ["Kankon", "Mondal", "kankonnil@gmail.com"]
fname, lname, email = x
print(fname, lname, email)
```
