# Dates and Times
This notebook covers working with dates and times in Python using the `datetime` and `time` modules.


```python
# Import datetime and time modules
import datetime as dt
import time as tm
```

```python
# Get the current time in seconds since the epoch
tm.time()
```

```python
# Convert the current timestamp to a datetime object
dtnow = dt.datetime.fromtimestamp(tm.time())
dtnow
```

```python
# Extract individual components of the datetime object
dtnow.year, dtnow.month, dtnow.day, dtnow.hour, dtnow.minute, dtnow.second
```

```python
# Create a timedelta object covering 100 days and find the date 100 days ago
delta = dt.timedelta(days=100)

today = dt.date.today()

today - delta # the date 100 days ago
```
